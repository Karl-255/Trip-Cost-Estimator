Trip Cost Estimator


```python
def trip_cost():
    """
    Calculate the estimated cost of a trip based on
    fuel, lodging, food, and other expenses.
    """
    distance = float(input("Distance (miles): "))
    mpg = float(input("Vehicle MPG: "))
    gas_price = float(input("Gas price per gallon: $"))

    lodging = float(input("Lodging cost: $"))
    food = float(input("Food cost: $"))
    other = float(input("Other expenses: $"))

    fuel_cost = (distance / mpg) * gas_price
    total_cost = fuel_cost + lodging + food + other

    print(f"\nFuel Cost: ${fuel_cost: .2f}")
    print(f"Total Trip Cost: ${total_cost: .2f}")

trip_cost()
```
